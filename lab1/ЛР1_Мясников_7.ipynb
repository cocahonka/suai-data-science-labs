### 3. Проверка пропусков

```python
# Посчитаем, сколько строк останется после удаления всех строк с пропусками
rows_after_dropna = df.dropna().shape[0]

# Исходное количество строк
initial_rows = df.shape[0]

# Количество удаленных строк
deleted_rows = initial_rows - rows_after_dropna

# Процент потерянных данных
loss_percentage = (deleted_rows / initial_rows) * 100

print(f"Исходное количество строк: {initial_rows}")
print(f"Количество строк после удаления пропусков: {rows_after_dropna}")
print(f"Будет удалено строк: {deleted_rows}")
print(f"Процент потерь: {loss_percentage:.2f}%")
```

```python
# Проверить данные на наличие пропусков и устранить их, если они есть (пропуски необходимо либо удалить, либо заменить каким-то значением).
print('Общее количество пропусков')
print(df.isna().sum())

# Замена аномальных значений на NaN
df.loc[df['age'] > 100, 'age'] = np.nan
df.loc[df['employment_length'] > 60, 'employment_length'] = np.nan

print("\nКоличество пропусков после замены аномалий:")
print(df.isna().sum())

print('\n Описательная статистика после замены аномалий:')
print(df.describe())

# Расчет медианных значений
median_age = df['age'].median()
median_employment_length = df['employment_length'].median()
median_rate = df['rate'].median()

# Заменяем все NaN на медианные значения
df = df.fillna({'age': median_age})
df = df.fillna({'employment_length': median_employment_length})
df = df.fillna({'rate': median_rate})

print("\nКоличество пропусков после замены NaN на медианные значения:")
print(df.isna().sum())
print('\n Описательная статистика после замены NaN на медианные значения:')
print(df.describe())
```

### Обоснование выбранной стратегии обработки пропусков

В ходе анализа данных были выявлены пропуски в столбцах `employment_length` и `loan_rate`, а также аномальные значения в `person_age` и `employment_length`, которые были преобразованы в пропуски.

**Выбранная стратегия:** Заполнение пропусков **медианным** значением.

**Аргументация:**

1.  **Сохранение данных:** Удаление строк с пропусками привело бы к потере значительной части данных (более 11%), что могло бы исказить результаты. Поэтому был выбран метод заполнения.
2.  **Надежность:** Для числовых признаков был выбран метод заполнения медианой, так как она, в отличие от среднего арифметического, устойчива к выбросам и экстремальным значениям. Это гарантирует, что пропуски заменяются наиболее "типичным" значением для большинства данных, а не смещенным из-за аномалий.

### Альтернативные способы работы с пропусками

1.  **Удаление строк или столбцов (`.dropna()`):** Простой метод, который подходит, если пропусков очень мало. В нашем случае это привело бы к значительной потере данных.
2.  **Заполнение средним значением (`.fillna(df['column'].mean())`):** Подходит для числовых данных без сильных выбросов. В нашем наборе данных этот метод был бы менее точным из-за наличия аномально высоких значений.
3.  **Заполнение нулем или другим значением (`.fillna(0)`):** Применяется, когда можно логически предположить, что пропуск означает отсутствие чего-либо (например, стажа).

## Дополинтельное задание №23

Отфильтровать набор данных. Выбрать топ 2 цели кредита по средней сумме
кредита + топ 2 статуса землевладения по медианной сумме кредита. Создать
сводные таблицы: цель кредита, статус домовладения - медианный, средний,
максимальный доход.

```python
# Топ-2 цели кредита по средней сумме
mean_amount_by_intent = df.groupby('intent')['amount'].mean().sort_values(ascending=False)
print("Средняя сумма кредита по целям:")
print(mean_amount_by_intent)

top_intents = mean_amount_by_intent.index[:2].tolist()
print(f"\nОтобранные топ-2 цели кредита: {top_intents}\n")


# Топ-2 статуса владения по медианной сумме
median_amount_by_ownership = df.groupby('home_ownership')['amount'].median().sort_values(ascending=False)
print("Медианная сумма кредита по статусу владения:")
print(median_amount_by_ownership)

top_ownerships = median_amount_by_ownership.index[:2].tolist()
print(f"\nОтобранные топ-2 статуса владения: {top_ownerships}")
```

```python
# Фильтруем датафрейм по найденным критериям из предыдущей ячейки
filtered_df = df[
    (df['intent'].isin(top_intents)) &
    (df['home_ownership'].isin(top_ownerships))
]

print('Отфильтрованный датафрейм (head):')
print(filtered_df.head())

# Итоговая сводная таблица
income_analysis_pivot = filtered_df.pivot_table(
    index='intent',
    columns='home_ownership',
    values='income',
    aggfunc=['median', 'mean', 'max'],
).round(2)
print('Итоговая сводная таблица:')
income_analysis_pivot
```

Анализ доходов топ-сегментов заемщиков (цели `DEBTCONSOLIDATION`, `PERSONAL` и статусы `MORTGAGE`, `RENT`) показал:

1.  **Заемщики с ипотекой (`MORTGAGE`) значительно состоятельнее.** Их медианный доход (**160 000 у.е.**) более чем в два раза превышает доход арендаторов, а максимальный доход достигает **500 000 у.е.**, что подтверждает их высокую платежеспособность.

2.  **Консолидация долга (`DEBTCONSOLIDATION`) — удел более обеспеченных.** Эта цель привлекает клиентов с более высоким средним доходом (**128 852 у.е.** у ипотечников) по сравнению с кредитами на личные цели. У арендаторов (`RENT`) такой ярко выраженной разницы в доходе между целями не наблюдается.
