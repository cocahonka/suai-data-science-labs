### 3. Проверка пропусков

```python
# Проверить данные на наличие пропусков и устранить их, если они есть 
# (пропуски необходимо либо удалить, либо заменить каким-то значением).

# Из анализа df.describe пропуски были у колонок:
# - employee_length
# - rate

na_df = df.isna().sum()
na_employee_length = df['employee_length'].isna().sum()
na_rate = df['rate'].isna().sum()

df_count = df.shape[0]

print('Общее количество пропусков по каждой колонке')
print(na_df)

print()
print('Суммарное количество пропусков в датасете')
print(na_df.sum())

print()
print('Количество пропусков в колонках employee_length и rate суммарно')
print(na_employee_length + na_rate)

print()
print('Процент пропусков в колонке employee_length')
print(na_employee_length / df_count * 100)

print()
print('Процент пропусков в колонке rate')
print(na_rate / df_count * 100)
```

```python
# Количество пропуском во всем датасете сошлось с количеством пропуском в колонках employee_length и rate,
# что и следовало ожидать.

# Так как процент пропусков по полю employee_length составляет всего ~1%,
# а разрброс значений не является большим (25 перцентиль - 2, а 75 перцентиль - 7)
# и дополнительно данное поле не является ключевым при анализе риска по кредитам,
# то лучшем решением будет заменить пропуски медианным значением.

df['employee_length'] = df['employee_length'].fillna(df['employee_length'].median())
print('Количество пропусков в employee_length после замены')
print(df['employee_length'].isna().sum())

# Так как количество пропусков по полю rate составляет около 11%, что довольно много, 
# а учитывая тот факт, что данное поле является важным при анализе риска по кредитам,
# то лучшем решением будет удалить строки, где rate пропущен.
df = df.dropna(subset=['rate'])
# Восстанавливаем индексацию при помощи reset_index.
# Не создаем столбец index, при помощи аргумента drop=True.
df = df.reset_index(drop=True)
print()
print('Количество пропусков в rate после удаления строк')
print(df['rate'].isna().sum())

print()
print('Количество пропусков в датасете после замены пропусков')
print(df.isna().sum().sum())

print()
print('Колечество строк в датафрейме')
print(df.shape[0])
```


---

Выводы по проверке и исправлению пропусков:
- Пропуски в данных были выявлены еще при анализе вывода describe(), дополинтельно подтвердили наличие пропусков с помощью isna().sum().
- Всего две колонки содержали пропуски, но обработаны были по разному.
- Пропуски в колонке employee_length были заменены медианным значением, а строки с пропусками в rate были удалены.
- Еще одним возможным решением для исправления пропусков является заполнение данных нулевым состоянием (0, '', None) или стандартным значением, если это применимо в конкретной ситуации - в данной ситауции это неприменимо.

### 4. Проверка дубликатов и аномалий в данных

#### Проверка явных дубликатов

```python
# Проверяем количество явных дубликатов при помощи duplicated
print('Количество явных дубликатов')
print(df.duplicated().sum())
```

```python
# Удаление явных дубликатов при помощи drop_duplicates
df = df.drop_duplicates()

# Восстанавливаем индексацию при помощи reset_index.
# Не создаем столбец index, при помощи аргумента drop=True.
df = df.reset_index(drop=True)

print('Колечество строк в датафрейме после удаления явных дубликатов')
print(df.shape[0])
```

#### Проверка неявных дубликатов

```python
# Неявные дубликаты могут содержаться в колонках с типом object, так как могут быть написаны неверно.
# В текущем датасете это:
# - home_ownership
# - intent
# - default

# Для поиска неявных дубликатов воспользуемся методом unique для нахождения всех уникальных значений в столбце.
print('Уникальные значения в столбце home_ownership')
print(df['home_ownership'].unique())

print()
print('Уникальные значения в столбце intent')
print(df['intent'].unique())

print()
print('Уникальные значения в столбце default')
print(df['default'].unique())

# Дополнительно выведем количество каждого уникального значения в столбце.
print()
print('Количество каждого уникального значения в столбце home_ownership')
print(df['home_ownership'].value_counts())

print()
print('Количество каждого уникального значения в столбце intent')
print(df['intent'].value_counts())

print()
print('Количество каждого уникального значения в столбце default')
print(df['default'].value_counts())
```

```python
# Удаление дубликатов

# В столбце intent отсутствуют дубликаты, а количество уникальных значений intent распределено нормально.
# Никаких действий не требуется.

# В столбце home_ownership, видно явную опечатку в виде RENET.
# Также в столбце присуствет уникальное значение OTHER, но всего в 2 строках,
# Так как 2 строки не играют роли в изучении рисков по кредитам, они будут удалены.

# Переименовываем RENET в RENT
df['home_ownership'] = df['home_ownership'].replace('RENET', 'RENT')

# Удаляем строки с OTHER и восстанавливаем индексацию
df = df[df['home_ownership'] != 'OTHER']
df = df.reset_index(drop=True)

# В столбце default наблюдается уникальный вариант No, что является аналогией к N.
# Переименовываем No в N
df['default'] = df['default'].replace('No', 'N')

print('Уникальные значения в столбце home_ownership')
print(df['home_ownership'].unique())

print()
print('Уникальные значения в столбце default')
print(df['default'].unique())

```

#### Проверка аномалий

```python
# Ранее в анализе были обнаружены следующие аномалии (не включая уже разрешенные)
# - слишком большое максимальное значение у employee_length
# - слишком большое максимальное значение у age
# - id дублирует стандартные индексы
# - status хоть целочисленный и по логике должен содержать 0 или 1, но возможно там есть и другие значения.

# Подготовим данные для изменений
# Найдем 5 максимальных значений employee_length,
# И если они будут нормальными, заменим аномалный employee_length на медиану.
# А если нет, то выставим барьер самостоятельно в некоторую константу.
print("5 максимальных значений employee_length")
employee_length_largest = df['employee_length'].nlargest(5)
print(employee_length_largest)

# Найдем 5 максимальных значений age по такой же логике.
print()
print("5 максимальных значений age")
age_largest = df['age'].nlargest(5)
print(age_largest)

# Посмотрим, есть ли дубликаты в id и если есть, выведем.
print()
print("Дубликаты в id")
print(df[df.duplicated('id')])

# Посмотрим на уникальные значения status
print()
print("Уникальные значения status")
print(df['status'].unique())
```

```python
# Аномалий в столбце id не обнаружено.
# Видимо в ранних обработках датасета, аномалия уже была убрана.
# Произойти это могло при удалении или фильтрации.

# Аномалий в status также не обнаружено.
# Все строки содержат сторого 1 или 0.

# Из вывода максимальных значений employee_length видно 2 выброса по значению 123.
# Заменим эти выбросы медианой.
# Замена в данном случае хороший вариант - так как мы сохраним данные, пусть и не много,
# а медиана не разрушит существующие тендендции.
# Дополнительно - стаж не такой важный параметр при расчете рисков по кредитам.
df['employee_length'] = df['employee_length'].replace(123, df['employee_length'].median())

# Из вывода по возрасту также видно 3 выброса
# - 144
# - 123
# Так как возраст также не важный параметр, заменим все значение >100 медианой.
# Чтобы не делать replace два раза, заменим значения >100 на NaN, а потом медианой.
# Замена будет производится при помощи loc.
df.loc[df['age'] > 100, 'age'] = np.nan
df['age'] = df['age'].fillna(df['age'].median())

print('После обработки')
print(df[['age', 'employee_length']].describe())
```

---

Выводы по дубликатам:
- Явные дубликаты были легко удалены при помощи drop_duplicates и drop_duplicates.
- Неявные дубликаты были найдены при помощи анализа unique, а затем удалены или изменены другими значениями.
- Дополнительно была удалена немногочисленная группа OTHER, которая ничего не дала бы для анализа.

Выводы по аномалиям:
- Все необработанные аномалии заключались в больших максимальных значениях и были изменены на медиану.

### 5. Провека типов данных

```python
# Проверьте типы данных, при необходимости измените типы данных, чтобы они соответствовали действительности.

# Из раннего анализа были выявлены следующие проблемы с типами
# - age имеет вещественный тип данных
# - employee_length тоже имеет вещественный тип данных
# - status имеет целочисленный тип данных
# - default имеет тип object

# Исправление типов данных

# Исправление для age
df['age'] = df['age'].astype('int64')

# Исправление для employee_length
df['employee_length'] = df['employee_length'].astype('int64')

# Исправление для status
df['status'] = df['status'].astype('bool')

# Исправление для default
df['default'] = (df['default'] == 'Y')

print('Типы данных после исправления:')
changed_types = df[['age', 'employee_length', 'status', 'default']]
print(changed_types.info())
print()
print(changed_types.head())
```

```python
# Группируем также по нескольким столбцам.
# Аналогично прошлому заданию применяем size()
# С помощью reset_index задаем имя столбцу с количеством.
# Сортируем по убыванию столбец с количеством.
home_by_default = df.groupby(['home_ownership', 'default']).size()
home_by_default = home_by_default.reset_index(name='count')
home_by_default = home_by_default.sort_values(by='count', ascending=False)
print('Датафрейм с количеством записей по каждой комбинации home_ownership и default')
print(home_by_default)

# Но возможно в описании задания "по количеству дефолтов" имелось ввиду только default=True,
# Поэтому выполним все то же самое, только предварительно отфильтровав данные по default=True.
home_by_default_true = df[df['default'] == True]
home_by_default_true = home_by_default_true.groupby(['home_ownership']).size()
home_by_default_true = home_by_default_true.reset_index(name='count')
home_by_default_true = home_by_default_true.sort_values(by='count', ascending=False)
print()
print('Датафрейм с количеством записей по home_ownership если default=True')
print(home_by_default_true)
```

#### Задание 3

Сводная таблица (pivot_table) - средняя сумма кредита (amount) для каждой цели кредита. Отсортировать по убыванию. Округлить до трёх знаков.

```python
# Создаем сводную таблицу.
# Индексом будет цель кредита - критерий по которому мы группируем.
# Значения - сумма кредита.
# Функция агрегации - среднее значение.
# aggfunc имеет стандартное значение mean, 
# но для лучшей ясности, явно указываю этот аргумент.
pivot = df.pivot_table(
    index='intent',
    values='amount',
    aggfunc='mean'
)
# Сортируем по сумме кредита в порядке убывания.
pivot = pivot.sort_values(by='amount', ascending=False)
# Округляем до 3 знаков после запятой.
pivot = pivot.round(3)
# Переименовываем столбец с суммой кредита в mean_amount для ясности.
pivot = pivot.rename(columns={'amount': 'mean_amount'})
pivot
```

Выводы из полученной таблицы:
- Наибольшую средную сумму кредита имеет цель погашения кредита (DEBTCONSOLIDATION). То есть люди берут кредит, чтобы погасить кредит. Но судить вероятно стоит по медиане, а не средней сумме, так как возможны выбросы.
- Вообще цели DEBTCONSOLIDATION, EDUCATION, MEDICAL и отчасти PERSONAL имеют почти одинковую среднюю сумму кредита, что говорит о их равной "популярности", если опять же доверять средней сумме, а не медиане.
- Наименьшую среднюю сумму кредита имеет цель улучшения дома (HOMEIMPROVEMENT).

```python
# Индекс - статус домовладения.
# Значения - доход.
# Функция агрегации - среднее значение.
# Сроки и цели кредита - столбцы.
# Непонятно, что имелось ввиду под "строки и цели кредита", вероятно "строки" = "сроки" (credit_length)
pivot = df.pivot_table(
    index='home_ownership',
    columns=['credit_length', 'intent'],
    values='income',
    aggfunc='mean'
)
# Сортируем по индексу (по возрастанию home_ownership)
pivot = pivot.sort_index()
# Округляем до 2 знаков после запятой
pivot = pivot.round(2)
pivot
```

#### Дополинтельное задание №2

Создать столбец "Категория дохода (Income)" с помощью категоризации, выделив 3 категории: низкий, средний и высокий доход. 

Фильтрацию границ категорий выбрать на основе описательной статистики данных. 

Затем построить сводную таблицу, показывающую средний и медианный доход для каждой категории в разрезе статуса домовладения (home_ownership).

```python
# Описательная статистика данных для дохода
df[['income']].describe()
```

```python
# Так как в задании сказано профильтровать категории на основе описательной статистики,
# То для 3х категории (низкий, средний, высокий) дохода возьмем следующие параметры:
# Низкий доход: 25% перцентиль (0 - 25%)
# Средний доход: 75% перцентиль (25% - 75%)
# Высокий доход: max (75% - 100%)

# Таким образом средний доход будет занимать 50% сегмента, 
# когда как низкий и высокий доход будут занимать 25%.

# Если бы не нужно было использовать описательную статистику,
# то было бы логичнее разделить на 3 ровных сегмента.
# 0-33% - низкий доход, 33-66% - средний доход, 66-100% - высокий доход
# Но в задании сказано профильтровать категории на основе описательной статистики, 
# поэтому будем использовать что дает описательная статистика.

stats = df['income'].describe()

# Выделяем пределы для каждой категории
low_income = stats['25%']
mid_income = stats['75%']
high_income = stats['max']

# Создаем новый столбец с категориями.
# Используем pd.cut для распределения диапазонов.
# Диапазоны будут начинаться с нуля, а не с low_income, что логично.
df['income_category'] = pd.cut(
    df['income'],
    bins=[0, low_income, mid_income, high_income], 
    labels=['low', 'mid', 'high'],
    include_lowest=True,
)

print('Выполненное разбиение на категории')
print(df[['income', 'income_category']])
print()
print(df.groupby('income_category', observed=True).size().sort_values())

# Создаем сводную таблицу.
# Индекс - категория дохода
# Разрез/столбец - статус домовладения.
# Значение - доход.
# Функции агрегации - среднее и медиана.
# observed = False - учитывать пропуски + 
# jupiter notebook ругается на отсутсвие флага observed в целом.
pivot = df.pivot_table(
    index='income_category',
    columns='home_ownership',
    values='income',
    aggfunc=['mean', 'median'],
    observed=False
)

pivot
```

Выводы по таблице и распределению:
- Как и предполагалось, из-за деления данных по описательной статистике (перцентилям), средняя группа занимает 50% данных, а низкая и высокая 25%.
- Люди со своим жильем не попадают в категорию "средний доход", только низкий или высокий.
- Самый высокий доход по среднему значению также у собственников жилья, но по медиане, самый высокий доход у ипотечников.
- Самый низкий доход по среднему и медианному значению также у собственников жилья.

Выбрать топ 2 цели кредита по средней сумме кредита + топ 2 статуса землевладения по медианной сумме кредита. 

Создать сводные таблицы: цель кредита, статус домовладения - медианный, средний, максимальный доход.

```python
# Группируем по цели кредита,
# вычисляем среднее значение суммы кредита для каждой цели.
# Сортируем результат по возрастанию.
# Берем два первых значения.
# Переводим индексы в python список.
top2intent = df.groupby('intent')['amount'].mean().sort_values().index[:2].to_list()

print('Топ 2 целей кредита')
print(top2intent)

# Аналогично вычисляем топ 2 статуса землевладения
top2home_ownership = df.groupby('home_ownership')['amount'].median().sort_values().index[:2].to_list()

print()
print('Топ 2 статуса землевладения')
print(top2home_ownership)
print()

# Фильтруем DataFrame по топ 2 целей кредита и статусу землевладения.
# Используем isin для проверки вхождения в список.
# Используем & для объединения условий.
filtered_df = df[
    df['intent'].isin(top2intent) &
    df['home_ownership'].isin(top2home_ownership)
]

pivot = filtered_df.pivot_table(
    index=['intent', 'home_ownership'],
    values='income',
    aggfunc=['median', 'mean', 'max'],
)
pivot
```
