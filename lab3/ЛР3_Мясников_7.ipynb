## 1.2. Модель линейной регрессии y от x1​ и x2

**Регрессия** — это класс моделей, задача которых состоит в том, чтобы по набору признаков (входных переменных) предсказать числовую целевую переменную. В отличие от классификации, где ответом является класс, в регрессии рассматриваются непрерывные значения (например, цена, зарплата, температура).

**Простая линейная регрессия** рассматривает зависимость одной числовой переменной y от одного признака x и предполагает, что связь между ними можно описать линейной функцией вида:

y = a * x + b,

где a — коэффициент наклона (показывает, как в среднем изменяется y при изменении x на единицу), а b — свободный член (значение y при x = 0).

Модель линейной регрессии подбирает такие параметры a и b, чтобы предсказанные значения (обозначим их как y_pred) были как можно ближе к истинным значениям y в смысле выбранной функции потерь (чаще всего используется минимизация суммы квадратов ошибок). На практике это позволяет по наблюдаемому значению признака x получать прогноз целевой переменной y и оценивать, насколько хорошо линейная модель описывает существующую зависимость в данных.


### 1.2.1. Формирование обучающей выборки (выделение признака и целевой переменной)

На данном шаге из исходного датафрейма выделяются отдельные матрицы признаков `X1`, `X2` и вектор целевой переменной `y`, а также выводятся их формы для контроля размеров.

```python
# формирование матриц признаков X1, X2 и вектора целевой переменной y
X1 = reg_df[['x1']]
X2 = reg_df[['x2']]
y = reg_df['y']

print('форма X1:', X1.shape)
print('форма X2:', X2.shape)
print('длина y:', y.shape)
```

### 1.2.1.1. Разделение данных на обучающую и валидационную выборки

Для корректной оценки качества моделей, как и в части 3, разобьём исходные данные `reg_df` на обучающую и валидационную выборки. 

Сначала сформируем общую матрицу признаков по `x1` и `x2`, затем с помощью `train_test_split` получим обучающий и валидационный поднаборы. 

Далее во всех моделях первой и второй части обучение будет выполняться на обучающей выборке, а метрики и сравнение — на валидационной.


```python
# разделение данных на обучающую и валидационную выборки для моделей y ~ x1 и y ~ x2

# общая матрица признаков по x1 и x2 и вектор целевой переменной
aX = reg_df[["x1", "x2"]]
ay = reg_df["y"]

X_train, X_val, y_train, y_val = train_test_split(
    aX,
    ay,
    test_size=0.2,  # 20% на валидацию
    random_state=77,
)

# отдельные матрицы признаков для x1 и x2 на train/validation
X1_train = X_train[["x1"]]
X2_train = X_train[["x2"]]

X1_val = X_val[["x1"]]
X2_val = X_val[["x2"]]

print("размер обучающей выборки:", X_train.shape[0])
print("размер обучающей выборки X1:", X1_train.shape[0])
print("размер обучающей выборки X2", X2_train.shape[0])
print()
print("размер валидационной выборки:", X_val.shape[0])
print("размер валидационной выборки X1:", X1_val.shape[0])
print("размер валидационной выборки X2", X2_val.shape[0])
```

### 1.2.2. Обучение модели простой линейной регрессии

На следующем этапе обучаются две модели простой линейной регрессии: первая предсказывает `y` по `x1`, вторая — по `x2`; после обучения выводятся найденные коэффициенты наклона и свободные члены для каждой модели.

```python
# обучение моделей простой линейной регрессии y от x1 и y от x2 на обучающей выборке
linreg_x1 = LinearRegression()
linreg_x2 = LinearRegression()

# модели обучаются только на train-части (как и в части 3)
# метод fit подберёт коэффициенты a и b в формуле y = a * x + b
linreg_x1.fit(X1_train, y_train)
linreg_x2.fit(X2_train, y_train)

# после обучения можно посмотреть значения коэффициентов модели
# coef_[0] — наклон (a), intercept_ — свободный член (b)
a_x1 = linreg_x1.coef_[0]
b_x1 = linreg_x1.intercept_

a_x2 = linreg_x2.coef_[0]
b_x2 = linreg_x2.intercept_

print('коэффициент наклона a (для x1):', a_x1)
print('свободный член b (для x1):', b_x1)
print()
print('коэффициент наклона a (для x2):', a_x2)
print('свободный член b (для x2):', b_x2)
```

### 1.2.3. Предсказание значений

```python
# получение предсказанных значений y_pred для валидационной выборки

y_pred_x1_val = linreg_x1.predict(X1_val)
y_pred_x2_val = linreg_x2.predict(X2_val)

print('первые 5 истинных значений y (валидация):')
print(y_val.head().values)
print()

print('первые 5 предсказанных значений y_pred (X1, валидация):')
print(y_pred_x1_val[:5])
print()

print('первые 5 предсказанных значений y_pred (X2, валидация):')
print(y_pred_x2_val[:5])

```

### 1.2.4. Формирование датафрейма с истинными и предсказанными значениями

```python
# формирование отдельного датафрейма для сравнения истинных и предсказанных значений (валидация)
results_x1 = pd.DataFrame({
    'x1': X1_val['x1'],
    'y_true': y_val,
    'y_pred': y_pred_x1_val,
})

results_x2 = pd.DataFrame({
    'x2': X2_val['x2'],
    'y_true': y_val,
    'y_pred': y_pred_x2_val,
})

print(results_x1.head())
print()
print(results_x2.head())
```

### 1.2.5. Расчёт метрик качества (MSE, MAE, RMSE, R²)

```python
# вычисление основных метрик качества для моделей y от x1 и y от x2 на валидационной выборке
mse_x1 = mean_squared_error(y_val, y_pred_x1_val)   # среднеквадратичная ошибка (MSE)
mse_x2 = mean_squared_error(y_val, y_pred_x2_val)
mae_x1 = mean_absolute_error(y_val, y_pred_x1_val)  # средняя абсолютная ошибка (MAE)
mae_x2 = mean_absolute_error(y_val, y_pred_x2_val)
rmse_x1 = np.sqrt(mse_x1)                           # корень из MSE (RMSE)
rmse_x2 = np.sqrt(mse_x2)
r2_x1 = r2_score(y_val, y_pred_x1_val)              # коэффициент детерминации R^2
r2_x2 = r2_score(y_val, y_pred_x2_val)

print('X1 (валидация)')
print('MSE  (среднеквадратичная ошибка):', mse_x1)
print('MAE  (средняя абсолютная ошибка):', mae_x1)
print('RMSE (корень из MSE):', rmse_x1)
print('R^2  (коэффициент детерминации):', r2_x1)
print()
print('X2 (валидация)')
print('MSE  (среднеквадратичная ошибка):', mse_x2)
print('MAE  (средняя абсолютная ошибка):', mae_x2)
print('RMSE (корень из MSE):', rmse_x2)
print('R^2  (коэффициент детерминации):', r2_x2)
```

### 1.2.8. График «истинные vs предсказанные» с диагональю 45°

Далее для каждой модели строится график зависимости предсказаний от истинных значений `y` с диагональной линией `y = x`, что позволяет визуально оценить точность предсказаний и разброс относительно идеального случая.

```python
# графики «истинные vs предсказанные» с диагональю 45° для обеих моделей (валидация)

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.scatter(y_val, y_pred_x1_val, alpha=0.6, label='предсказания модели (x1), валидация')
min_val = min(y_val.min(), y_pred_x1_val.min())
max_val = max(y_val.max(), y_pred_x1_val.max())
plt.plot([min_val, max_val], [min_val, max_val], color='red', label='идеальные предсказания (y = x)')
plt.title('Сравнение истинных и предсказанных значений (y от x1), валидация')
plt.xlabel('истинные значения y (валидация)')
plt.ylabel('предсказанные значения y_pred_x1')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.scatter(y_val, y_pred_x2_val, alpha=0.6, label='предсказания модели (x2), валидация')
min_val = min(y_val.min(), y_pred_x2_val.min())
max_val = max(y_val.max(), y_pred_x2_val.max())
plt.plot([min_val, max_val], [min_val, max_val], color='red', label='идеальные предсказания (y = x)')
plt.title('Сравнение истинных и предсказанных значений (y от x2), валидация')
plt.xlabel('истинные значения y (валидация)')
plt.ylabel('предсказанные значения y_pred_x2')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

## 1.3. Сравнение моделей по x1 и x2 и выводы

- По всем метрикам ошибок на валидационной выборке модель y ~ x1 даёт меньшие значения MSE, MAE и RMSE по сравнению с моделью y ~ x2.
- Коэффициент детерминации R² на валидации также выше у y ~ x1, то есть первая модель лучше объясняет вариацию целевой переменной на «невидимых» данных.
- На графике «истинные vs предсказанные» для y ~ x1 облако точек ближе к диагонали y = x, тогда как для y ~ x2 разброс относительно диагонали больше.
- Линии регрессии у обеих моделей восходящие (положительная связь), но линия для y ~ x1 проходит ближе к основной массе точек валидационной выборки, что согласуется с меньшими ошибками.
- Для простой линейной регрессии на данном наборе данных признак x1 является более информативным, чем x2; для дальнейшего улучшения можно пробовать более сложные виды регрессии, но оценивать их качество также нужно по валидационной выборке.

# 2. Полиномиальная регрессия

## 2.1. Анализ целесообразности выбора признака

Для полиномиальной регрессии целесообразно использовать x1. Линейная модель y ~ x1 показала более высокое качество на валидационной выборке и компактные остатки вокруг нуля. Это означает, что базовая линейная зависимость уже сильная. Добавление полиномиального члена позволяет проверить наличие слабой кривизны на краях диапазона x1. Для x2 линейная связь слабее, и добавление полинома рискует дать избыточные колебания без устойчивого прироста качества, поэтому в работе полином строится только по x1.

## 2.2. Подготовка данных и генерация полиномиальных признаков

На этом этапе выбираются конкретные степени полинома для построения моделей. В качестве базовой нелинейной аппроксимации используется степень 2 (квадратичный полином), которая позволяет уловить простую кривизну зависимости `y` от `x1`. Для проверки, даёт ли усложнение дополнительный прирост качества, добавляется ещё одна модель со степенью 3.

Дополнительно рассматривается более высокая степень 20, чтобы показать, как усложнение полинома может приводить к подстройке под шум и росту риска переобучения. Теоретически можно использовать и другие высокие степени, однако при фиксированном объёме и наличии шума такие модели начинают сильнее подгоняться под случайные колебания: кривая становится «изломанной», возрастает риск переобучения и ухудшается обобщающая способность. Поэтому в работе сравниваются три степени (2, 3 и 20) что позволяет продемонстрировать принцип: сначала выбирается невысокая степень, затем постепенно добавляются более сложные модели и оценивается баланс между приростом качества и риском переобучения.

### 2.2.1. Формирование полиномиальных признаков

Далее формируются полиномиальные признаки по признаку `x1` для трёх степеней полинома (2, 3 и 20) с помощью `PolynomialFeatures`. На их основе создаются расширенные матрицы признаков, которые используются при обучении моделей полиномиальной регрессии и последующем расчёте метрик качества.


```python
# формирование полиномиальных признаков для x1 при трёх фиксированных степенях полинома

# полиномиальные признаки для степени 2 (квадратичный полином)
poly2 = PolynomialFeatures(degree=2, include_bias=False)
X1_poly_deg2_train = poly2.fit_transform(X1_train)
X1_poly_deg2_val = poly2.transform(X1_val)

# полиномиальные признаки для степени 3 (кубический полином)
poly3 = PolynomialFeatures(degree=3, include_bias=False)
X1_poly_deg3_train = poly3.fit_transform(X1_train)
X1_poly_deg3_val = poly3.transform(X1_val)

# полиномиальные признаки для степени 20 (полином высокой степени)
poly20 = PolynomialFeatures(degree=20, include_bias=False)
X1_poly_deg20_train = poly20.fit_transform(X1_train)
X1_poly_deg20_val = poly20.transform(X1_val)

# проверка размеров сформированных матриц признаков (train и validation)
print(f"степень полинома 2: X1_poly_deg2_train = {X1_poly_deg2_train.shape}, X1_poly_deg2_val = {X1_poly_deg2_val.shape}")
print(f"степень полинома 3: X1_poly_deg3_train = {X1_poly_deg3_train.shape}, X1_poly_deg3_val = {X1_poly_deg3_val.shape}")
print(f"степень полинома 20: X1_poly_deg20_train = {X1_poly_deg20_train.shape}, X1_poly_deg20_val = {X1_poly_deg20_val.shape}")
```

### 2.2.2. Обучение моделей полиномиальной регрессии

После формирования полиномиальных признаков обучаются три модели полиномиальной регрессии для степеней 2, 3 и 20 на матрицах `X1_poly_deg2`, `X1_poly_deg3` и `X1_poly_deg20`. Для каждой степени создаётся и обучается отдельный объект `LinearRegression`, что позволяет в дальнейшем сопоставить качество аппроксимации между моделями.


```python
# обучение моделей полиномиальной регрессии для степеней 2, 3 и 20 (на обучающей выборке)

linreg_poly_deg2 = LinearRegression()
linreg_poly_deg3 = LinearRegression()
linreg_poly_deg20 = LinearRegression()

linreg_poly_deg2.fit(X1_poly_deg2_train, y_train)
linreg_poly_deg3.fit(X1_poly_deg3_train, y_train)
linreg_poly_deg20.fit(X1_poly_deg20_train, y_train)

print(f"степень полинома 2: форма вектора коэффициентов = {linreg_poly_deg2.coef_.shape}, интерсепт = {linreg_poly_deg2.intercept_:.4f}")
print(f"степень полинома 3: форма вектора коэффициентов = {linreg_poly_deg3.coef_.shape}, интерсепт = {linreg_poly_deg3.intercept_:.4f}")
print(f"степень полинома 20: форма вектора коэффициентов = {linreg_poly_deg20.coef_.shape}, интерсепт = {linreg_poly_deg20.intercept_:.4f}")
```

### 2.2.3. Предсказание и расчёт метрик

Затем для каждой модели полиномиальной регрессии вычисляются предсказания `y_pred` на исходных данных и рассчитываются метрики качества `MAE` и `R²`. Полученные значения сводятся в отдельную таблицу, что позволяет сравнить степени полинома (2, 3 и 20) между собой и с линейной моделью из первой части.

```python
# предсказания и метрики качества для степеней 2, 3 и 20 (на валидационной выборке)

y_pred_poly_deg2_val = linreg_poly_deg2.predict(X1_poly_deg2_val)
y_pred_poly_deg3_val = linreg_poly_deg3.predict(X1_poly_deg3_val)
y_pred_poly_deg20_val = linreg_poly_deg20.predict(X1_poly_deg20_val)

mae_deg2 = mean_absolute_error(y_val, y_pred_poly_deg2_val)
r2_deg2 = r2_score(y_val, y_pred_poly_deg2_val)

mae_deg3 = mean_absolute_error(y_val, y_pred_poly_deg3_val)
r2_deg3 = r2_score(y_val, y_pred_poly_deg3_val)

mae_deg20 = mean_absolute_error(y_val, y_pred_poly_deg20_val)
r2_deg20 = r2_score(y_val, y_pred_poly_deg20_val)

print(f"степень 2 (валидация): MAE = {mae_deg2:.4f}, R² = {r2_deg2:.4f}")
print(f"степень 3 (валидация): MAE = {mae_deg3:.4f}, R² = {r2_deg3:.4f}")
print(f"степень 20 (валидация): MAE = {mae_deg20:.4f}, R² = {r2_deg20:.4f}")

# сводная таблица по степеням полинома
metrics_poly_df = pd.DataFrame({
    'degree': [2, 3, 20],
    'MAE_val': [mae_deg2, mae_deg3, mae_deg20],
    'R2_val': [r2_deg2, r2_deg3, r2_deg20],
})

print("\nсводная таблица метрик для полиномиальных моделей (валидация):")
print(metrics_poly_df)
```

Для наглядности исходные точки `(x1, y)` отображаются в виде диаграммы рассеяния, поверх которой наносится кривая полиномиальной регрессии для степеней 2 и 3. Сравнение форм кривых и их прилегания к данным позволяет дополнительно оценить качество аппроксимации и возможные признаки переобучения.

При этом на графиках кривая степени 2 остаётся практически линейной, степень 3 даёт чуть более гибкую, но всё ещё гладкую аппроксимацию, а степень 20 начинает заметно повторять мелкие колебания облака точек.

С практической точки зрения такой выигрыш в метриках при переходе к степени 20 невелик и сопровождается усложнением формы кривой, что является признаком начала переобучения: модель подстраивается под шум и может хуже обобщать данные вне наблюдаемого диапазона. Степень 3 даёт небольшой прирост качества по сравнению с 2 при сохранении гладкой зависимости, поэтому в рамках работы разумно считать оптимальной невысокую степень (2–3), а степени порядка 20 рассматривать как демонстрацию того, как чрезмерное увеличение сложности модели приводит к рискам переобучения.

В общем случае выбор степени полинома выполняется на основе совокупности численных метрик и визуального анализа графиков: слишком низкая степень приводит к недообучению и систематическим ошибкам, а слишком высокая — к переобучению и чувствительности к шуму. Оптимальная степень должна обеспечивать заметное улучшение `MAE`/`R²` без излишнего усложнения формы регрессионной кривой.


## 3. Решение задачи регрессии различными методами

### 3.1. Загрузка набора данных

В данном пункте считывается файл `salary.csv` в датафрейм `sal_df`, выполняется первичный просмотр нескольких строк и вывод общей информации о столбцах, типах данных и размере набора.

```python
# чтение salary.csv и первичная проверка структуры набора

sal_df = pd.read_csv('salary.csv')

print('размерность набора:', sal_df.shape)
print('\nпервые строки:')
print(sal_df.head(5))
print('\nинформация о столбцах:')
print(sal_df.info())
```

### 3.2. Целевая переменная — визуализация (гистограмма и boxplot)

На этом шаге выбирается целевая переменная `salary`, после чего строятся гистограмма и boxplot по этому столбцу, чтобы оценить форму распределения и наличие возможных выбросов.

```python
# выбор целевой переменной salary и визуализация её распределения

target = 'salary'

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# гистограмма целевой переменной
sb.histplot(sal_df[target], ax=axes[0])
axes[0].set_title('гистограмма: salary')
axes[0].set_xlabel('salary')
axes[0].set_ylabel('частота')

# boxplot целевой переменной
sb.boxplot(x=sal_df[target], orient='h', width=0.3, ax=axes[1])
axes[1].set_title('boxplot: salary')
axes[1].set_xlabel('salary')
axes[1].set_yticks([])

plt.tight_layout()
plt.show()
```

### 3.3. Обработка выбросов и повторная визуализация

В пункте выполняется выявление выбросов по целевой переменной `salary` с помощью правила IQR, формируется очищенный датафрейм `sal_df_cln`, после чего повторно строятся гистограмма и boxplot для оценки влияния удаления выбросов на распределение.

```python
# обработка выбросов по правилу IQR и повторная визуализация salary

q1 = sal_df[target].quantile(0.25)
q3 = sal_df[target].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

sal_df_cln = sal_df[(sal_df[target] >= lower) & (sal_df[target] <= upper)].copy()

removed = sal_df.shape[0] - sal_df_cln.shape[0]
print(f'удалено по IQR: {removed} из {sal_df.shape[0]} записей')
print('новая размерность:', sal_df_cln.shape)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# гистограмма после удаления выбросов
sb.histplot(sal_df_cln[target], bins='auto', ax=axes[0])
axes[0].set_title('гистограмма: salary (после очистки)')
axes[0].set_xlabel('salary')
axes[0].set_ylabel('частота')

# boxplot после удаления выбросов
sb.boxplot(x=sal_df_cln[target], orient='h', width=0.3, ax=axes[1])
axes[1].set_title('boxplot: salary (после очистки)')
axes[1].set_xlabel('salary')
axes[1].set_yticks([])

plt.tight_layout()
plt.show()
```

```python
# построение матрицы диаграмм рассеяния для числовых признаков

numeric_cols = sal_df_cln.select_dtypes(include='number').columns.tolist()
print('числовые столбцы:', numeric_cols)

sb.pairplot(sal_df_cln[numeric_cols])
plt.suptitle('матрица диаграмм рассеяния (числовые признаки)', y=1.02)
plt.show()
```

### 3.5. Разделение на обучающую и тестовую выборки

На данном шаге данные разделяются на обучающую и тестовую выборки: формируется матрица признаков `X` (все столбцы, кроме `salary`) и вектор целей `y`, после чего используется `train_test_split` для получения `X_train`, `X_test`, `y_train`, `y_test`.

```python
# матрица признаков X и вектор целевой переменной y_full
X = sal_df_cln.drop(columns=[target])
y_full = sal_df_cln[target]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_full,
    test_size=0.2, # 20% для теста
    random_state=77, # зерно рандома
)

print('размер X_train:', X_train.shape)
print('размер X_test :', X_test.shape)
print('размер y_train:', y_train.shape)
print('размер y_test :', y_test.shape)
```

### 3.6. Масштабирование числовых признаков

Здесь выполняется масштабирование числовых признаков в обучающей и тестовой выборках с помощью `StandardScaler`: числовые столбцы стандартизируются (обнуляется среднее и нормируется дисперсия), категориальные признаки остаются без изменений и будут кодироваться позднее.

```python
# выделение числовых и категориальных столбцов
numeric_features = X_train.select_dtypes(include='number').columns.tolist()
cat_features = X_train.select_dtypes(exclude='number').columns.tolist()

print('числовые признаки:', numeric_features)
print('категориальные признаки:', cat_features)

scaler = StandardScaler()

# обучение StandardScaler на обучающей выборке и преобразование train/test
X_train_scaled_num = scaler.fit_transform(X_train[numeric_features])
X_test_scaled_num = scaler.transform(X_test[numeric_features])

# сборка обратно в датафреймы с сохранением категориальных признаков
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numeric_features] = X_train_scaled_num
X_test_scaled[numeric_features] = X_test_scaled_num

print('\nразмер X_train_scaled:', X_train_scaled.shape)
print('размер X_test_scaled :', X_test_scaled.shape)
```

### 3.7. Линейная регрессия — обучение

В этом пункте подготавливаются признаки для линейной регрессии: категориальные столбцы кодируются с помощью one-hot-представления на обучающей выборке и затем применяется обученная модель `LinearRegression` для обучения на `X_train` и `y_train`.

```python
# one-hot кодирование категориальных признаков на обучающей и тестовой выборках
# pd.get_dummies создаёт по одному бинарному столбцу на каждую категорию (one-hot представление),
# параметр drop_first=True опускает один из столбцов для каждой группы категорий,
# чтобы избежать линейной зависимости признаков (ловушки дамми-переменных).
X_train_enc = pd.get_dummies(X_train_scaled, columns=cat_features, drop_first=True)
X_test_enc = pd.get_dummies(X_test_scaled, columns=cat_features, drop_first=True)

# выравнивание столбцов тестовой выборки под обучающую (на случай отсутствующих категорий)
X_test_enc = X_test_enc.reindex(columns=X_train_enc.columns, fill_value=0)

print('размер X_train_enc:', X_train_enc.shape)
print('размер X_test_enc :', X_test_enc.shape)

# обучение модели линейной регрессии
linreg_main = LinearRegression()
linreg_main.fit(X_train_enc, y_train)

print('обучение линейной регрессии завершено')
```

### 3.8. Линейная регрессия — оценка на тесте

Здесь обученная модель линейной регрессии применяется к тестовой выборке, после чего рассчитываются метрики качества (MSE, MAE, RMSE, R²) для оценки точности предсказаний.

```python
# применение обученной модели к тестовой выборке и расчёт метрик качества

y_pred_test = linreg_main.predict(X_test_enc)

mse_test = mean_squared_error(y_test, y_pred_test)
mae_test = mean_absolute_error(y_test, y_pred_test)
rmse_test = np.sqrt(mse_test)
r2_test = r2_score(y_test, y_pred_test)

print('MSE  (среднеквадратичная ошибка):', mse_test)
print('MAE  (средняя абсолютная ошибка):', mae_test)
print('RMSE (корень из MSE):', rmse_test)
print('R^2  (коэффициент детерминации):', r2_test)
```

### 3.9. Таблица с истинными и предсказанными значениями

В данном пункте формируется отдельный датафрейм с истинными значениями `salary` на тестовой выборке и соответствующими предсказаниями модели, что позволяет наглядно сопоставить результаты и использовать таблицу для дальнейшего анализа.

```python
# формирование датафрейма с истинными и предсказанными значениями на тесте

results_test = pd.DataFrame({
    'y_true': y_test.values,
    'y_pred': y_pred_test,
})

print(results_test.head())
```

### 3.10. Важность признаков — таблица коэффициентов линейной модели

На заключительном шаге для линейной регрессии формируется таблица коэффициентов по всем признакам: по величине и знаку коэффициентов оценивается вклад каждого признака в итоговый прогноз `salary`, после чего делаются выводы о наиболее и наименее значимых факторах.

На этом шаге модель k-ближайших соседей применяется к тестовой выборке, рассчитываются те же метрики качества, что и для линейной регрессии, формируется сравнительная таблица и строятся графики для визуального сравнения двух подходов.

Дерево решений строит последовательность бинарных разбиений признакового пространства: на каждом узле выбирается признак и числовой порог, по которому выборка делится на две части так, чтобы уменьшить разброс целевой переменной (обычно по критерию среднеквадратичной ошибки). в листьях дерева хранятся средние значения `salary` для объектов, попавших в лист, и именно они используются как предсказания для новых примеров.

Для дерева решений дополнительно подбирается значение основного гиперпараметра `max_depth` (глубина дерева), после чего качество лучшей найденной конфигурации оценивается на тестовой выборке `X_test_enc`, `y_test` по тем же метрикам (MSE, MAE, RMSE, R²), что и для линейной регрессии и kNN.

```python
# реализация регрессии методом дерева решений на закодированных признаках

# перебор нескольких значений глубины дерева и выбор лучшего по R^2 на тестовой выборке
max_depth_values = [None, 3, 5, 7, 10]

depth_results = []

for depth in max_depth_values:
    tree_tmp = DecisionTreeRegressor(max_depth=depth, random_state=77)
    tree_tmp.fit(X_train_enc, y_train)
    y_pred_tmp = tree_tmp.predict(X_test_enc)

    mse_tmp = mean_squared_error(y_test, y_pred_tmp)
    mae_tmp = mean_absolute_error(y_test, y_pred_tmp)
    rmse_tmp = np.sqrt(mse_tmp)
    r2_tmp = r2_score(y_test, y_pred_tmp)

    depth_results.append({
        'max_depth': depth,
        'MSE': mse_tmp,
        'MAE': mae_tmp,
        'RMSE': rmse_tmp,
        'R^2': r2_tmp,
    })

depth_results_df = pd.DataFrame(depth_results)
print('качество дерева решений при разных значениях max_depth (оценка на тестовой выборке):')
print(depth_results_df)

# выбор глубины с наибольшим значением R^2
best_depth = int(depth_results_df.loc[depth_results_df['R^2'].idxmax(), 'max_depth'])
print('\nвыбранное значение max_depth для итоговой модели дерева решений:', best_depth)

# итоговая модель дерева решений с подобранной глубиной
reg_tree_base = DecisionTreeRegressor(max_depth=best_depth, random_state=77)
reg_tree_base.fit(X_train_enc, y_train)

# предсказания на тестовой выборке для итоговой модели
y_pred_tree_base = reg_tree_base.predict(X_test_enc)

# расчёт метрик качества для итоговой модели дерева решений
mse_tree_base = mean_squared_error(y_test, y_pred_tree_base)
mae_tree_base = mean_absolute_error(y_test, y_pred_tree_base)
rmse_tree_base = np.sqrt(mse_tree_base)
r2_tree_base = r2_score(y_test, y_pred_tree_base)

print('\nDecisionTreeRegressor (модель с подобранным max_depth):')
print('MSE  :', mse_tree_base)
print('MAE  :', mae_tree_base)
print('RMSE :', rmse_tree_base)
print('R^2  :', r2_tree_base)

# сводная таблица метрик для трёх моделей: линейная регрессия, kNN и дерево решений
metrics_compare_tree = pd.DataFrame({
    'MSE':  [mse_test, mse_knn, mse_tree_base],
    'MAE':  [mae_test, mae_knn, mae_tree_base],
    'RMSE': [rmse_test, rmse_knn, rmse_tree_base],
    'R^2':  [r2_test, r2_knn, r2_tree_base],
}, index=['LinearRegression', 'KNN (k=5)', 'DecisionTree (best_depth)'])

print('\nсравнение метрик моделей:')
print(metrics_compare_tree)

# scatter-график для итоговой модели дерева решений
plt.figure(figsize=(14, 4))

# отдельный график только для дерева решений
plt.subplot(1, 2, 1)
min_val_tree = min(y_test.min(), y_pred_tree_base.min())
max_val_tree = max(y_test.max(), y_pred_tree_base.max())

plt.scatter(y_test.values, y_pred_tree_base, alpha=0.6, label='DecisionTree (best_depth)')
plt.plot([min_val_tree, max_val_tree], [min_val_tree, max_val_tree], color='red', label='идеальные предсказания (y = x)')
plt.title('DecisionTreeRegressor (best_depth): y_true vs y_pred (test)')
plt.xlabel('y_true')
plt.ylabel('y_pred_tree_base')
plt.legend()
plt.grid(True)

# общий график для сравнения всех трёх моделей на одних и тех же точках
plt.subplot(1, 2, 2)
min_val_all = min(y_test.min(), y_pred_test.min(), y_pred_knn.min(), y_pred_tree_base.min())
max_val_all = max(y_test.max(), y_pred_test.max(), y_pred_knn.max(), y_pred_tree_base.max())

plt.plot([min_val_all, max_val_all], [min_val_all, max_val_all], color='black', label='идеальные предсказания (y = x)')
plt.scatter(y_test.values, y_pred_test, alpha=0.5, label='LinearRegression', color='tab:blue')
plt.scatter(y_test.values, y_pred_knn, alpha=0.5, label='KNN (k=5)', color='tab:orange')
plt.scatter(y_test.values, y_pred_tree_base, alpha=0.5, label='DecisionTree (best_depth)', color='tab:green')
plt.title('сравнение моделей: y_true vs y_pred (test)')
plt.xlabel('y_true (salary)')
plt.ylabel('y_pred')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

### 4.2. Выводы по задаче дерева решений

После подбора гиперпараметра max_depth на тестовой выборке наилучшее качество дерево решений показало при глубине 5: r² ≈ 0.66 и rmse ≈ 8.9k. по всем метрикам такая модель заметно лучше, чем исходное «глубокое» дерево (без ограничения глубины) и kNN‑регрессия (k=5), но всё ещё существенно уступает линейной регрессии, у которой r² ≈ 0.85 и заметно ниже mae и rmse. на scatter‑графиках видно, что точки дерева с max_depth=5 лежат ближе к диагонали y = x, чем у kNN, однако линия идеальных предсказаний всё равно обложена облаком с большим разбросом, чем в случае линейной модели. 

В итоге дерево решений после простого подбора глубины становится приемлемой альтернативой knn, но для данного набора данных лучшим по качеству и стабильности остаётся линейная регрессия.

### 7.1. Презумпции по подбору гиперпараметров дерева решений

В этом пункте для модели дерева решений (`DecisionTreeRegressor`) выполняется подбор гиперпараметров с помощью `GridSearchCV`.

Для поиска используются разумные диапазоны значений `max_depth`, `min_samples_split` и `min_samples_leaf`, качество оценивается по метрике R² с кросс-валидацией на обучающей выборке. Лучшая найденная конфигурация затем переобучается на всей train-выборке и сравнивается по тестовым метрикам с базовым деревом решений и ранее реализованными моделями.

```python
# подбор гиперпараметров дерева решений с помощью GridSearchCV

# базовый объект дерева решений для поиска по сетке
reg_tree_for_grid = DecisionTreeRegressor(random_state=77)

# сетка гиперпараметров: глубина дерева и размеры узлов
param_grid_tree = {
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
}

# GridSearchCV с оценкой по R^2 и 5-кратной кросс-валидацией
tree_grid_search = GridSearchCV(
    estimator=reg_tree_for_grid,
    param_grid=param_grid_tree,
    scoring='r2',
    cv=5,
    n_jobs=-1,
)

tree_grid_search.fit(X_train_enc, y_train)

print('лучшие параметры дерева решений (по R^2 на кросс-валидации):')
print(tree_grid_search.best_params_)
print('\nлучшее среднее значение R^2 на CV:', tree_grid_search.best_score_)

# лучшая модель по результатам поиска
best_tree_model = tree_grid_search.best_estimator_

# оценка лучшей модели на тестовой выборке
y_pred_tree_best = best_tree_model.predict(X_test_enc)

mse_tree_best = mean_squared_error(y_test, y_pred_tree_best)
mae_tree_best = mean_absolute_error(y_test, y_pred_tree_best)
rmse_tree_best = np.sqrt(mse_tree_best)
r2_tree_best = r2_score(y_test, y_pred_tree_best)

print('\nDecisionTreeRegressor (модель после GridSearchCV):')
print('MSE  :', mse_tree_best)
print('MAE  :', mae_tree_best)
print('RMSE :', rmse_tree_best)
print('R^2  :', r2_tree_best)

# расширенная сравнительная таблица метрик для всех моделей
metrics_compare_all = pd.DataFrame({
    'MSE':  [mse_test, mse_knn, mse_tree_base, mse_tree_best],
    'MAE':  [mae_test, mae_knn, mae_tree_base, mae_tree_best],
    'RMSE': [rmse_test, rmse_knn, rmse_tree_base, rmse_tree_best],
    'R^2':  [r2_test, r2_knn, r2_tree_base, r2_tree_best],
}, index=['LinearRegression', 'KNN (k=5)', 'DecisionTree (best_depth)', 'DecisionTree (GridSearchCV)'])

print('\nсводная таблица метрик для всех реализованных моделей:')
print(metrics_compare_all)

# визуализация: сравнение базового и улучшенного дерева решений на scatter-графике
plt.figure(figsize=(6, 4))

min_val = min(y_test.min(), y_pred_tree_base.min(), y_pred_tree_best.min())
max_val = max(y_test.max(), y_pred_tree_base.max(), y_pred_tree_best.max())


# базовая модель дерева решений (синие точки)
plt.scatter(
    y_test.values,
    y_pred_tree_base,
    alpha=1,
    label='DecisionTree (base)',
    color='tab:blue',
)

# модель после GridSearchCV (оранжевые точки)
plt.scatter(
    y_test.values,
    y_pred_tree_best,
    alpha=0.5,
    label='DecisionTree (GridSearchCV)',
    color='tab:orange',
)


plt.plot([min_val, max_val], [min_val, max_val], color='red', label='идеальные предсказания (y = x)')
plt.title('базовое дерево решений и модель после GridSearchCV: y_true vs y_pred (test)')
plt.xlabel('y_true (salary)')
plt.ylabel('y_pred')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

при использовании GridSearchCV для дерева решений был автоматически подобран набор гиперпараметров `max_depth = 5`, `min_samples_split = 10`, `min_samples_leaf = 1`, причём среднее значение r² на кросс‑валидации составило ≈ 0.67, а итоговое качество на тестовой выборке осталось тем же, что и у модели с подобранной глубиной: r² ≈ 0.66 и rmse ≈ 8.9k. по всем метрикам такое дерево сильно выигрывает у исходной базовой конфигурации и у kNN, но всё ещё заметно уступает линейной регрессии, которая остаётся лучшей моделью по точности. 

GridSearchCV здесь играет роль "автоматизированного перебора" гиперпараметров: вместо ручного перебора одного параметра, как в задании 4, он перебирает целую сетку значений (глубину, минимальный размер узла и листа), оценивая каждую комбинацию по r² на кросс‑валидации и выбирая наилучшую, тогда как сами коэффициенты/разбиения дерева по‑прежнему обучаются обычным алгоритмом.
