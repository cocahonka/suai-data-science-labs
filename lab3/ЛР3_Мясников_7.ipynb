## 1.2. Модель линейной регрессии y от x1​ и x2

**Регрессия** — это класс моделей, задача которых состоит в том, чтобы по набору признаков (входных переменных) предсказать числовую целевую переменную. В отличие от классификации, где ответом является класс, в регрессии рассматриваются непрерывные значения (например, цена, зарплата, температура).

**Простая линейная регрессия** рассматривает зависимость одной числовой переменной y от одного признака x и предполагает, что связь между ними можно описать линейной функцией вида:

y = a * x + b,

где a — коэффициент наклона (показывает, как в среднем изменяется y при изменении x на единицу), а b — свободный член (значение y при x = 0).

Модель линейной регрессии подбирает такие параметры a и b, чтобы предсказанные значения (обозначим их как y_pred) были как можно ближе к истинным значениям y в смысле выбранной функции потерь (чаще всего используется минимизация суммы квадратов ошибок). На практике это позволяет по наблюдаемому значению признака x получать прогноз целевой переменной y и оценивать, насколько хорошо линейная модель описывает существующую зависимость в данных.


### 1.2.1. Формирование обучающей выборки (выделение признака и целевой переменной)

На данном шаге из исходного датафрейма выделяются отдельные матрицы признаков `X1`, `X2` и вектор целевой переменной `y`, а также выводятся их формы для контроля размеров.

```python
# формирование матриц признаков X1, X2 и вектора целевой переменной y
X1 = reg_df[['x1']]
X2 = reg_df[['x2']]
y = reg_df['y']

print('форма X1:', X1.shape)
print('форма X2:', X2.shape)
print('длина y:', y.shape)
```

### 1.2.2. Обучение модели простой линейной регрессии

На следующем этапе обучаются две модели простой линейной регрессии: первая предсказывает `y` по `x1`, вторая — по `x2`; после обучения выводятся найденные коэффициенты наклона и свободные члены для каждой модели.

```python
# обучение моделей простой линейной регрессии y от x1 и y от x2
linreg_x1 = LinearRegression()
linreg_x2 = LinearRegression()

# обучаем модель на всех доступных наблюдениях (X1, y) (X2, y)
# метод fit подберёт коэффициенты a и b в формуле y = a * x1 + b
linreg_x1.fit(X1, y)
linreg_x2.fit(X2, y)

# после обучения можно посмотреть значения коэффициентов модели
# coef_[0] — наклон (a), intercept_ — свободный член (b)
a_x1 = linreg_x1.coef_[0]
b_x1 = linreg_x1.intercept_

a_x2 = linreg_x2.coef_[0]
b_x2 = linreg_x2.intercept_

print('коэффициент наклона a (для x1):', a_x1)
print('свободный член b (для x1):', b_x1)
print()
print('коэффициент наклона a (для x2):', a_x2)
print('свободный член b (для x2):', b_x2)
```

### 1.2.3. Предсказание значений

```python
# получение предсказанных значений y_pred для тех же наблюдений, что использовались при обучении

y_pred_x1 = linreg_x1.predict(X1)
y_pred_x2 = linreg_x2.predict(X2)

print('первые 5 истинных значений y:')
print(y.head().values)
print()

print('первые 5 предсказанных значений y_pred (X1):')
print(y_pred_x1[:5])
print()

print('первые 5 предсказанных значений y_pred (X2):')
print(y_pred_x2[:5])

```

### 1.2.4. Формирование датафрейма с истинными и предсказанными значениями

```python
# формирование отдельного датафрейма для сравнения истинных и предсказанных значений
results_x1 = pd.DataFrame({
    'x1': X1['x1'],
    'y_true': y,
    'y_pred': y_pred_x1,
})

results_x2 = pd.DataFrame({
    'x2': X2['x2'],
    'y_true': y,
    'y_pred': y_pred_x2,
})

print(results_x1.head())
print()
print(results_x2.head())
```

### 1.2.5. Расчёт метрик качества (MSE, MAE, RMSE, R²)

```python
# вычисление основных метрик качества для модели y от x1 b x2
mse_x1 = mean_squared_error(y, y_pred_x1)   # среднеквадратичная ошибка (MSE)
mse_x2 = mean_squared_error(y, y_pred_x2) 
mae_x1 = mean_absolute_error(y, y_pred_x1)  # средняя абсолютная ошибка (MAE)
mae_x2 = mean_absolute_error(y, y_pred_x2)
rmse_x1 = np.sqrt(mse_x1)                   # корень из MSE (RMSE)
rmse_x2 = np.sqrt(mse_x2)
r2_x1 = r2_score(y, y_pred_x1)              # коэффициент детерминации R^2
r2_x2 = r2_score(y, y_pred_x2)

print('X1')
print('MSE  (среднеквадратичная ошибка):', mse_x1)
print('MAE  (средняя абсолютная ошибка):', mae_x1)
print('RMSE (корень из MSE):', rmse_x1)
print('R^2  (коэффициент детерминации):', r2_x1)
print()
print('X2')
print('MSE  (среднеквадратичная ошибка):', mse_x2)
print('MAE  (средняя абсолютная ошибка):', mae_x2)
print('RMSE (корень из MSE):', rmse_x2)
print('R^2  (коэффициент детерминации):', r2_x2)
```

### 1.2.6. Визуализация: scatter + линия регрессии

```python
# Визуализация: точки (scatter plot) и линия регрессии (y от x1)

plt.figure(figsize=(14, 5))

# линия регрессии: для аккуратного отображения отсортируем x1 по возрастанию
# и посчитаем соответствующие предсказания
x1_sorted = np.sort(X1['x1'].values)
x1_sorted_df = pd.DataFrame({'x1': x1_sorted})

x2_sorted = np.sort(X2['x2'].values)
x2_sorted_df = pd.DataFrame({'x2': x2_sorted})

y_pred_line_x1 = linreg_x1.predict(x1_sorted_df)
y_pred_line_x2 = linreg_x2.predict(x2_sorted_df)

plt.subplot(1, 2, 1)
plt.scatter(X1['x1'], y, label='истинные значения', alpha=0.6)
plt.plot(x1_sorted, y_pred_line_x1, color='red', label='линия регрессии')
plt.title('Модель линейной регрессии: y от x1')
plt.xlabel('x1')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.scatter(X2['x2'], y, label='истинные значения', alpha=0.6)
plt.plot(x2_sorted, y_pred_line_x2, color='red', label='линия регрессии')
plt.title('Модель линейной регрессии: y от x2')
plt.xlabel('x2')
plt.ylabel('y')
plt.legend()
plt.grid(True)


plt.tight_layout()
plt.show()
```

### 1.2.7. Визуализация ошибок по точкам (разница истинное–предсказанное)

```python
# визуализация ошибок (y_true - y_pred) для моделей y от x1 и y от x2

errors_x1 = y - y_pred_x1
errors_x2 = y - y_pred_x2

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.plot(errors_x1.values, marker='o', linestyle='none', alpha=0.6)
plt.axhline(0, color='red', linewidth=1)
plt.title('Ошибки модели (y_true - y_pred_x1) для регрессии y от x1')
plt.xlabel('номер наблюдения')
plt.ylabel('ошибка')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(errors_x2.values, marker='o', linestyle='none', alpha=0.6)
plt.axhline(0, color='red', linewidth=1)
plt.title('Ошибки модели (y_true - y_pred_x2) для регрессии y от x2')
plt.xlabel('номер наблюдения')
plt.ylabel('ошибка')
plt.grid(True)

plt.tight_layout()
plt.show()
```

### 1.2.8. График «истинные vs предсказанные» с диагональю 45°

```python
# графики «истинные vs предсказанные» с диагональю 45° для обеих моделей

plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.scatter(y, y_pred_x1, alpha=0.6, label='предсказания модели (x1)')
min_val = min(y.min(), y_pred_x1.min())
max_val = max(y.max(), y_pred_x1.max())
plt.plot([min_val, max_val], [min_val, max_val], color='red', label='идеальные предсказания (y = x)')
plt.title('Сравнение истинных и предсказанных значений (y от x1)')
plt.xlabel('истинные значения y')
plt.ylabel('предсказанные значения y_pred_x1')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.scatter(y, y_pred_x2, alpha=0.6, label='предсказания модели (x2)')
min_val = min(y.min(), y_pred_x2.min())
max_val = max(y.max(), y_pred_x2.max())
plt.plot([min_val, max_val], [min_val, max_val], color='red', label='идеальные предсказания (y = x)')
plt.title('Сравнение истинных и предсказанных значений (y от x2)')
plt.xlabel('истинные значения y')
plt.ylabel('предсказанные значения y_pred_x2')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

- Для простой линейной регрессии признак x1 является более сильным чем x2. Для дальнейшего улучшения имеет смысл попробовать другие виды регрессии.

# 2. Полиномиальная регрессия

## 2.1. Анализ целесообразности выбора признака

Для полиномиальной регрессии целесообразно использовать x1. Линейная модель y ~ x1 показала более высокое качество (R² ≈ 0.77) и компактные остатки вокруг нуля. Это означает, что базовая линейная зависимость уже сильная. Добавление полиномиального члена позволит проверить наличие слабой кривизны на краях диапазона x1. Для x2 линейная связь слабее (R² ≈ 0.62), и добавление полинома рискует дать избыточные колебания без устойчивого прироста качества.

## 2.2. Подготовка данных и генерация полиномиальных признаков

На этом этапе выбираются конкретные степени полинома для построения моделей. В качестве базовой нелинейной аппроксимации используется степень 2 (квадратичный полином), которая позволяет уловить простую кривизну зависимости `y` от `x1`. Для проверки, даёт ли усложнение дополнительный прирост качества, добавляется ещё одна модель со степенью 3.

Дополнительно рассматривается более высокая степень 20, чтобы показать, как усложнение полинома может приводить к подстройке под шум и росту риска переобучения. Теоретически можно использовать и другие высокие степени, однако при фиксированном объёме и наличии шума такие модели начинают сильнее подгоняться под случайные колебания: кривая становится «изломанной», возрастает риск переобучения и ухудшается обобщающая способность. Поэтому в работе сравниваются три степени (2, 3 и 20) что позволяет продемонстрировать принцип: сначала выбирается невысокая степень, затем постепенно добавляются более сложные модели и оценивается баланс между приростом качества и риском переобучения.

### 2.2.1. Формирование полиномиальных признаков

Далее формируются полиномиальные признаки по признаку `x1` для трёх степеней полинома (2, 3 и 20) с помощью `PolynomialFeatures`. На их основе создаются расширенные матрицы признаков, которые используются при обучении моделей полиномиальной регрессии и последующем расчёте метрик качества.


```python
# формирование полиномиальных признаков для x1 при трёх фиксированных степенях полинома

# полиномиальные признаки для степени 2 (квадратичный полином)
poly2 = PolynomialFeatures(degree=2, include_bias=False)
X1_poly_deg2 = poly2.fit_transform(X1)

# полиномиальные признаки для степени 3 (кубический полином)
poly3 = PolynomialFeatures(degree=3, include_bias=False)
X1_poly_deg3 = poly3.fit_transform(X1)

# полиномиальные признаки для степени 20 (полином высокой степени)
poly20 = PolynomialFeatures(degree=20, include_bias=False)
X1_poly_deg20 = poly20.fit_transform(X1)

# проверка размеров сформированных матриц признаков
print(f"степень полинома 2: форма X1_poly_deg2 = {X1_poly_deg2.shape}")
print(f"степень полинома 3: форма X1_poly_deg3 = {X1_poly_deg3.shape}")
print(f"степень полинома 20: форма X1_poly_deg20 = {X1_poly_deg20.shape}")
```

### 2.2.2. Обучение моделей полиномиальной регрессии

После формирования полиномиальных признаков обучаются три модели полиномиальной регрессии для степеней 2, 3 и 20 на матрицах `X1_poly_deg2`, `X1_poly_deg3` и `X1_poly_deg20`. Для каждой степени создаётся и обучается отдельный объект `LinearRegression`, что позволяет в дальнейшем сопоставить качество аппроксимации между моделями.


```python
# обучение моделей полиномиальной регрессии для степеней 2, 3 и 20

linreg_poly_deg2 = LinearRegression()
linreg_poly_deg3 = LinearRegression()
linreg_poly_deg20 = LinearRegression()

linreg_poly_deg2.fit(X1_poly_deg2, y)
linreg_poly_deg3.fit(X1_poly_deg3, y)
linreg_poly_deg20.fit(X1_poly_deg20, y)

print(f"степень полинома 2: форма вектора коэффициентов = {linreg_poly_deg2.coef_.shape}, интерсепт = {linreg_poly_deg2.intercept_:.4f}")
print(f"степень полинома 3: форма вектора коэффициентов = {linreg_poly_deg3.coef_.shape}, интерсепт = {linreg_poly_deg3.intercept_:.4f}")
print(f"степень полинома 20: форма вектора коэффициентов = {linreg_poly_deg20.coef_.shape}, интерсепт = {linreg_poly_deg20.intercept_:.4f}")
```

### 2.2.3. Предсказание и расчёт метрик

Затем для каждой модели полиномиальной регрессии вычисляются предсказания `y_pred` на исходных данных и рассчитываются метрики качества `MAE` и `R²`. Полученные значения сводятся в отдельную таблицу, что позволяет сравнить степени полинома (2, 3 и 20) между собой и с линейной моделью из первой части.

```python
# предсказания и метрики качества для степеней 2, 3 и 20

y_pred_poly_deg2 = linreg_poly_deg2.predict(X1_poly_deg2)
y_pred_poly_deg3 = linreg_poly_deg3.predict(X1_poly_deg3)
y_pred_poly_deg20 = linreg_poly_deg20.predict(X1_poly_deg20)

mae_deg2 = mean_absolute_error(y, y_pred_poly_deg2)
r2_deg2 = r2_score(y, y_pred_poly_deg2)

mae_deg3 = mean_absolute_error(y, y_pred_poly_deg3)
r2_deg3 = r2_score(y, y_pred_poly_deg3)

mae_deg20 = mean_absolute_error(y, y_pred_poly_deg20)
r2_deg20 = r2_score(y, y_pred_poly_deg20)

print(f"степень 2: MAE = {mae_deg2:.4f}, R² = {r2_deg2:.4f}")
print(f"степень 3: MAE = {mae_deg3:.4f}, R² = {r2_deg3:.4f}")
print(f"степень 20: MAE = {mae_deg20:.4f}, R² = {r2_deg20:.4f}")

# сводная таблица по степеням полинома
metrics_poly_df = pd.DataFrame({
    'degree': [2, 3, 20],
    'MAE': [mae_deg2, mae_deg3, mae_deg20],
    'R2': [r2_deg2, r2_deg3, r2_deg20],
})

print("\nсводная таблица метрик для полиномиальных моделей:")
print(metrics_poly_df)
```

### 2.2.4. Визуализация: точки и кривая полиномиальной регрессии

Для наглядности исходные точки `(x1, y)` отображаются в виде диаграммы рассеяния, поверх которой наносится кривая полиномиальной регрессии для степеней 2 и 3. Сравнение форм кривых и их прилегания к данным позволяет дополнительно оценить качество аппроксимации и возможные признаки переобучения.


```python
# визуализация полиномиальной регрессии для степеней 2, 3 и 20

plt.figure(figsize=(21, 5))

# подготовка упорядоченной сетки по x1 для построения гладких кривых
x1_grid = np.linspace(X1['x1'].min(), X1['x1'].max(), 200)

# формирование датафрейма с тем же именем столбца, что и у исходного признака
x1_grid_df = pd.DataFrame({'x1': x1_grid})

# формирование полиномиальных признаков для сетки и получение предсказаний
X1_grid_poly_deg2 = poly2.transform(x1_grid_df)
X1_grid_poly_deg3 = poly3.transform(x1_grid_df)
X1_grid_poly_deg20 = poly20.transform(x1_grid_df)

y_grid_deg2 = linreg_poly_deg2.predict(X1_grid_poly_deg2)
y_grid_deg3 = linreg_poly_deg3.predict(X1_grid_poly_deg3)
y_grid_deg20 = linreg_poly_deg20.predict(X1_grid_poly_deg20)

# график для степени 2
plt.subplot(1, 3, 1)
plt.scatter(X1['x1'], y, alpha=0.5, label='истинные значения')
plt.plot(x1_grid, y_grid_deg2, color='red', label='полиномиальная регрессия (degree=2)')
plt.title('Полиномиальная регрессия y от x1 (degree=2)')
plt.xlabel('x1')
plt.ylabel('y')
plt.legend()
plt.grid(True)

# график для степени 3
plt.subplot(1, 3, 2)
plt.scatter(X1['x1'], y, alpha=0.5, label='истинные значения')
plt.plot(x1_grid, y_grid_deg3, color='green', label='полиномиальная регрессия (degree=3)')
plt.title('Полиномиальная регрессия y от x1 (degree=3)')
plt.xlabel('x1')
plt.ylabel('y')
plt.legend()
plt.grid(True)

# график для степени 20
plt.subplot(1, 3, 3)
plt.scatter(X1['x1'], y, alpha=0.5, label='истинные значения')
plt.plot(x1_grid, y_grid_deg20, color='purple', label='полиномиальная регрессия (degree=20)')
plt.title('Полиномиальная регрессия y от x1 (degree=20)')
plt.xlabel('x1')
plt.ylabel('y')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

## 2.3. Сравнение степеней полинома и выбор оптимальной

По рассчитанным метрикам `MAE` и `R²` для степеней 2, 3 и 20 наблюдается небольшой численный выигрыш при усложнении модели: `MAE` уменьшается с ≈0.5327 (degree=2) до ≈0.5308 (degree=3) и ≈0.5256 (degree=20), а `R²` растёт с ≈0.7683 до ≈0.7690 и ≈0.7726 соответственно. При этом на графиках кривая степени 2 остаётся практически линейной, степень 3 даёт чуть более гибкую, но всё ещё гладкую аппроксимацию, а степень 20 начинает заметно повторять мелкие колебания облака точек.

С практической точки зрения такой выигрыш в метриках при переходе к степени 20 невелик и сопровождается усложнением формы кривой, что является признаком начала переобучения: модель подстраивается под шум и может хуже обобщать данные вне наблюдаемого диапазона. Степень 3 даёт небольшой прирост качества по сравнению с 2 при сохранении гладкой зависимости, поэтому в рамках работы разумно считать оптимальной невысокую степень (2–3), а степени порядка 20 рассматривать как демонстрацию того, как чрезмерное увеличение сложности модели приводит к рискам переобучения.

В общем случае выбор степени полинома выполняется на основе совокупности численных метрик и визуального анализа графиков: слишком низкая степень приводит к недообучению и систематическим ошибкам, а слишком высокая — к переобучению и чувствительности к шуму. Оптимальная степень должна обеспечивать заметное улучшение `MAE`/`R²` без излишнего усложнения формы регрессионной кривой.


## 3. Решение задачи регрессии различными методами

### 3.1. Загрузка набора данных

В данном пункте считывается файл `salary.csv` в датафрейм `sal_df`, выполняется первичный просмотр нескольких строк и вывод общей информации о столбцах, типах данных и размере набора.

```python
# чтение salary.csv и первичная проверка структуры набора

sal_df = pd.read_csv('salary.csv')

print('размерность набора:', sal_df.shape)
print('\nпервые строки:')
print(sal_df.head(5))
print('\nинформация о столбцах:')
print(sal_df.info())
```

### 3.2. Целевая переменная — визуализация (гистограмма и boxplot)

На этом шаге выбирается целевая переменная `salary`, после чего строятся гистограмма и boxplot по этому столбцу, чтобы оценить форму распределения и наличие возможных выбросов.

```python
# выбор целевой переменной salary и визуализация её распределения

target = 'salary'

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# гистограмма целевой переменной
sb.histplot(sal_df[target], ax=axes[0])
axes[0].set_title('гистограмма: salary')
axes[0].set_xlabel('salary')
axes[0].set_ylabel('частота')

# boxplot целевой переменной
sb.boxplot(x=sal_df[target], orient='h', width=0.3, ax=axes[1])
axes[1].set_title('boxplot: salary')
axes[1].set_xlabel('salary')
axes[1].set_yticks([])

plt.tight_layout()
plt.show()
```

### 3.3. Обработка выбросов и повторная визуализация

В пункте выполняется выявление выбросов по целевой переменной `salary` с помощью правила IQR, формируется очищенный датафрейм `sal_df_cln`, после чего повторно строятся гистограмма и boxplot для оценки влияния удаления выбросов на распределение.

```python
# обработка выбросов по правилу IQR и повторная визуализация salary

q1 = sal_df[target].quantile(0.25)
q3 = sal_df[target].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

sal_df_cln = sal_df[(sal_df[target] >= lower) & (sal_df[target] <= upper)].copy()

removed = sal_df.shape[0] - sal_df_cln.shape[0]
print(f'удалено по IQR: {removed} из {sal_df.shape[0]} записей')
print('новая размерность:', sal_df_cln.shape)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# гистограмма после удаления выбросов
sb.histplot(sal_df_cln[target], bins='auto', ax=axes[0])
axes[0].set_title('гистограмма: salary (после очистки)')
axes[0].set_xlabel('salary')
axes[0].set_ylabel('частота')

# boxplot после удаления выбросов
sb.boxplot(x=sal_df_cln[target], orient='h', width=0.3, ax=axes[1])
axes[1].set_title('boxplot: salary (после очистки)')
axes[1].set_xlabel('salary')
axes[1].set_yticks([])

plt.tight_layout()
plt.show()
```

### 3.4. Матрица диаграмм рассеяния по числовым признакам

Здесь формируется список числовых столбцов из очищенного набора `sal_df_cln` и строится матрица диаграмм рассеяния (pairplot), позволяющая визуально оценить взаимосвязи между числовыми признаками и целевой переменной `salary`.

```python
# построение матрицы диаграмм рассеяния для числовых признаков

numeric_cols = sal_df_cln.select_dtypes(include='number').columns.tolist()
print('числовые столбцы:', numeric_cols)

sb.pairplot(sal_df_cln[numeric_cols])
plt.suptitle('матрица диаграмм рассеяния (числовые признаки)', y=1.02)
plt.show()
```

### 3.5. Разделение на обучающую и тестовую выборки

На данном шаге данные разделяются на обучающую и тестовую выборки: формируется матрица признаков `X` (все столбцы, кроме `salary`) и вектор целей `y`, после чего используется `train_test_split` для получения `X_train`, `X_test`, `y_train`, `y_test`.

```python
# матрица признаков X и вектор целевой переменной y_full
X = sal_df_cln.drop(columns=[target])
y_full = sal_df_cln[target]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_full,
    test_size=0.2, # 20% для теста
    random_state=77, # зерно рандома
)

print('размер X_train:', X_train.shape)
print('размер X_test :', X_test.shape)
print('размер y_train:', y_train.shape)
print('размер y_test :', y_test.shape)
```

### 3.6. Масштабирование числовых признаков

Здесь выполняется масштабирование числовых признаков в обучающей и тестовой выборках с помощью `StandardScaler`: числовые столбцы стандартизируются (обнуляется среднее и нормируется дисперсия), категориальные признаки остаются без изменений и будут кодироваться позднее.

```python
# выделение числовых и категориальных столбцов
numeric_features = X_train.select_dtypes(include='number').columns.tolist()
cat_features = X_train.select_dtypes(exclude='number').columns.tolist()

print('числовые признаки:', numeric_features)
print('категориальные признаки:', cat_features)

scaler = StandardScaler()

# обучение StandardScaler на обучающей выборке и преобразование train/test
X_train_scaled_num = scaler.fit_transform(X_train[numeric_features])
X_test_scaled_num = scaler.transform(X_test[numeric_features])

# сборка обратно в датафреймы с сохранением категориальных признаков
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numeric_features] = X_train_scaled_num
X_test_scaled[numeric_features] = X_test_scaled_num

print('\nразмер X_train_scaled:', X_train_scaled.shape)
print('размер X_test_scaled :', X_test_scaled.shape)
```

### 3.7. Линейная регрессия — обучение

В этом пункте подготавливаются признаки для линейной регрессии: категориальные столбцы кодируются с помощью one-hot-представления на обучающей выборке и затем применяется обученная модель `LinearRegression` для обучения на `X_train` и `y_train`.

```python
# one-hot кодирование категориальных признаков на обучающей и тестовой выборках
# pd.get_dummies создаёт по одному бинарному столбцу на каждую категорию (one-hot представление),
# параметр drop_first=True опускает один из столбцов для каждой группы категорий,
# чтобы избежать линейной зависимости признаков (ловушки дамми-переменных).
X_train_enc = pd.get_dummies(X_train_scaled, columns=cat_features, drop_first=True)
X_test_enc = pd.get_dummies(X_test_scaled, columns=cat_features, drop_first=True)

# выравнивание столбцов тестовой выборки под обучающую (на случай отсутствующих категорий)
X_test_enc = X_test_enc.reindex(columns=X_train_enc.columns, fill_value=0)

print('размер X_train_enc:', X_train_enc.shape)
print('размер X_test_enc :', X_test_enc.shape)

# обучение модели линейной регрессии
linreg_main = LinearRegression()
linreg_main.fit(X_train_enc, y_train)

print('обучение линейной регрессии завершено')
```

### 3.8. Линейная регрессия — оценка на тесте

Здесь обученная модель линейной регрессии применяется к тестовой выборке, после чего рассчитываются метрики качества (MSE, MAE, RMSE, R²) для оценки точности предсказаний.

```python
# применение обученной модели к тестовой выборке и расчёт метрик качества

y_pred_test = linreg_main.predict(X_test_enc)

mse_test = mean_squared_error(y_test, y_pred_test)
mae_test = mean_absolute_error(y_test, y_pred_test)
rmse_test = np.sqrt(mse_test)
r2_test = r2_score(y_test, y_pred_test)

print('MSE  (среднеквадратичная ошибка):', mse_test)
print('MAE  (средняя абсолютная ошибка):', mae_test)
print('RMSE (корень из MSE):', rmse_test)
print('R^2  (коэффициент детерминации):', r2_test)
```

### 3.9. Таблица с истинными и предсказанными значениями

В данном пункте формируется отдельный датафрейм с истинными значениями `salary` на тестовой выборке и соответствующими предсказаниями модели, что позволяет наглядно сопоставить результаты и использовать таблицу для дальнейшего анализа.

### 3.10. Важность признаков — таблица коэффициентов линейной модели

На этом шаге модель k-ближайших соседей применяется к тестовой выборке, рассчитываются те же метрики качества, что и для линейной регрессии, формируется сравнительная таблица и строятся графики для визуального сравнения двух подходов.

```python
# оценка качества KNN-регрессии и сравнение с линейной моделью

# предсказания KNN на тестовой выборке
y_pred_knn = knn_reg.predict(X_test_enc)

# метрики качества для KNN
mse_knn = mean_squared_error(y_test, y_pred_knn)
mae_knn = mean_absolute_error(y_test, y_pred_knn)
rmse_knn = np.sqrt(mse_knn)
r2_knn = r2_score(y_test, y_pred_knn)

print('KNN-регрессия (k=5):')
print('MSE  :', mse_knn)
print('MAE  :', mae_knn)
print('RMSE :', rmse_knn)
print('R^2  :', r2_knn)

# сравнительная таблица метрик для линейной регрессии и KNN
metrics_compare = pd.DataFrame({
    'MSE':  [mse_test, mse_knn],
    'MAE':  [mae_test, mae_knn],
    'RMSE': [rmse_test, rmse_knn],
    'R^2':  [r2_test, r2_knn],
}, index=['LinearRegression', 'KNN (k=5)'])

print('\nсравнение метрик линейной регрессии и KNN:')
print(metrics_compare)

# визуальное сравнение: scatter-графики y_true vs y_pred для обеих моделей
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.scatter(y_test.values, y_pred_test, alpha=0.6, label='LinearRegression')
min_val = min(y_test.min(), y_pred_test.min())
max_val = max(y_test.max(), y_pred_test.max())
plt.plot([min_val, max_val], [min_val, max_val], color='red', label='идеальные предсказания (y = x)')
plt.title('LinearRegression: y_true vs y_pred (test)')
plt.xlabel('y_true')
plt.ylabel('y_pred')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.scatter(y_test.values, y_pred_knn, alpha=0.6, label='KNN (k=5)')
min_val = min(y_test.min(), y_pred_knn.min())
max_val = max(y_test.max(), y_pred_knn.max())
plt.plot([min_val, max_val], [min_val, max_val], color='red', label='идеальные предсказания (y = x)')
plt.title('KNN: y_true vs y_pred (test)')
plt.xlabel('y_true')
plt.ylabel('y_pred')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
